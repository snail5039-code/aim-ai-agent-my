### 문제 1
추상 클래스(`Abstract Base Class`)를 사용하여 도형의 면적을 계산하는 구조를 작성하세요.
1. `abc` 모듈에서 `ABC`와 `abstractmethod`를 임포트합니다.
2. `Shape`라는 추상 클래스를 만들고, `get_area(self)`라는 추상 메서드를 정의하세요.
3. `Shape`를 상속받는 `Square`(변의 길이 `side` 속성) 클래스를 만드세요.
4. `Square` 클래스에서 `get_area()`를 오버라이딩하여 정사각형의 면적을 반환하도록 구현하세요.

```python
# 여기에 코드를 작성하세요.
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod  
    def get_area(self):
        pass

class Square(Shape):
    def __init__(self,side : float):
        self.side : float = side
    def get_area(self) -> float:
        return self.side * self.side

square = Square(5)
square.get_area()
```

### 문제 2

### 문제 3
합성(`Composition`) 관계를 활용하여 컴퓨터와 CPU의 관계를 구현하세요.
1. `CPU` 클래스를 정의하고, 모델명(`model`)을 속성으로 가집니다.
2. `Computer` 클래스를 정의하고, 생성자(`__init__`)에서 `CPU` 객체를 인자로 받아 인스턴스 변수에 할당(Composition)하세요.
3. `Computer` 클래스에 `show_info(self)` 메서드를 만들어, 장착된 CPU의 모델명을 포함하여 "컴퓨터 사양: [CPU 모델명]"을 출력하세요.

```python
# 여기에 코드를 작성하세요.
# 여기에 코드를 작성하세요.

class CPU:
    def __init__(self, model : str):
        self.model : str = model

class Computer:
    def __init__(self, cpu : CPU):
        self.cpu : CPU = cpu

    def show_info(self):
        print(f'컴퓨터 사양 : {self.cpu.model}')
        

cpu = CPU("i5-1999")
computer = Computer(cpu)

computer.show_info()
```
