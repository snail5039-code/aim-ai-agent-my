## requests에서 사용하는 HTTP 메서드

`requests`는 HTTP 요청을 보내기 위한 파이썬 라이브러리이다. API의 목적에 따라 다음 메서드를 사용한다.

- `requests.get(url)`: 서버에서 데이터나 목록을 조회하는 메서드이다.
- `requests.post(url, json=data)`: 서버에 새로운 데이터를 생성하는 메서드이다.
- `requests.patch(url, json=data)`: 기존 데이터 중 필요한 일부만 수정하는 메서드이다.
- `requests.put(url, json=data)`: 기존 데이터를 새로운 내용으로 전체 교체하는 메서드이다.
- `requests.delete(url)`: 서버의 데이터를 삭제하는 메서드이다.

### 요청 옵션

- `params`: URL 뒤에 쿼리 매개변수를 추가할 때 사용한다. 예를 들어 `requests.get(url, params={"userId": 2})`는 `?userId=2`를 URL에 추가한다.
- `headers`: 인증 정보, 응답 형식 등 요청의 메타데이터를 전달할 때 사용한다. 예를 들어 `headers={"Authorization": "Bearer <token>"}`과 같이 인증 토큰을 전달한다.

각 메서드는 `Response` 객체를 반환한다. 응답 본문은 `response.json()`으로 확인하고, 요청 성공 여부는 `response.status_code` 또는 `response.raise_for_status()`로 확인한다.

또는 `response.text`로도 본문 확인이 가능함. 단, 이는 str이다.

### 자원 조회 (GET)
특정 데이터를 가져오거나 전체 목록을 불러올 때 사용한다.

```python
# 단일 리소스 조회
url = "https://jsonplaceholder.typicode.com/posts/10"
response = requests.get(url)
# print(\"단일 조회 결과:\", response.json())

data = response.json()
pprint(data)
print(data['body'])
```

```python

# 전체 목록 조회
list_url = "https://jsonplaceholder.typicode.com/posts"
list_response = requests.get(list_url)
print("전체 게시글 개수:", len(list_response.json()))
# pprint(list_response.json())
```

### 자원 생성 (POST)
새로운 데이터를 서버에 등록한다. 데이터는 JSON 형식으로 전달된다.

```python
url = "https://jsonplaceholder.typicode.com/posts"

new_data = {
    "title": "하하 호호123",
    "body": "새로운 내용",
    "userId": 1
}

response = requests.post(url, json=new_data)
print("생성 성공 상태 코드:", response.status_code)
print("응답 데이터:", response.json())
```

### 자원 수정 (PATCH)
데이터의 일부를 변경을 요청한다.

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

# 일부 수정
patch_data = {"title": "수정된 제목"}
patch_response = requests.patch(url, json=patch_data)
print("수정 결과:", patch_response.json())
```

### 전체 자원 수정 (PUT)
기존 자원의 모든 필드를 새 데이터로 교체한다.

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

# 전체 수정
put_data = {
    "title": "PUT으로 수정된 제목",
    "body": "PUT 요청으로 전체 내용을 교체합니다.",
    "userId": 1
}

put_response = requests.put(url, json=put_data)
print("전체 수정 상태 코드:", put_response.status_code)
print("수정 결과:", put_response.json())
```

### 자원 삭제 (DELETE)
데이터 삭제를 요청한다.

```python
url = "https://jsonplaceholder.typicode.com/posts/1"


# 삭제
delete_response = requests.delete(url)
print("삭제 성공 상태 코드:", delete_response.status_code)
```

### 필터링 (params)
보조적인 조건을 추가하여 정보를 가져온다.

```python
# 쿼리 매개변수를 활용한 필터링
# query_url = "https://jsonplaceholder.typicode.com/posts?userId=2"
# filter_response = requests.get(query_url)

query_url = "https://jsonplaceholder.typicode.com/posts"
params = {"userId": 2}
filter_response = requests.get(query_url, params=params)

print("사용자 1의 게시글 수:", len(filter_response.json()))
pprint(filter_response.json())
```

### 중첩 리소스 조회 (GET)
특정 자원에 종속된 정보를 가져온다.

```python
# 중첩 리소스 조회
nested_url = "https://jsonplaceholder.typicode.com/posts/1/comments"
nested_response = requests.get(nested_url)
print("1번 게시글의 첫 번째 댓글:", nested_response.json()[0])
pprint(nested_response.json())
```

### 예외 처리
요청 중 발생할 수 있는 네트워크 오류와 HTTP 오류를 처리한다.

- `Timeout`: 설정한 시간 안에 서버 응답이 오지 않을 때 발생한다.
- `HTTPError`: 서버에는 연결됐지만 404(자원 없음), 401(인증 필요), 500(서버 오류)처럼 4xx나 5xx 상태 코드를 응답할 때 발생한다.
- `raise_for_status()`: 응답 상태 코드가 200~399면 아무 일도 하지 않는다. 400대나 500대면 `HTTPError`를 발생시켜 `except`문에서 처리할 수 있게 한다. 호출하지 않으면 500이어도 예외는 발생하지 않고, `response.status_code`에 500이 담긴다.
- `RequestException`: 위 예외들을 포함한 요청 관련 오류의 상위 예외다. 별도로 분기할 필요가 없는 오류를 처리한다.


```python
# 정상
error_url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    response = requests.get(error_url, timeout=5)
    response.raise_for_status()
    print("요청 성공:", response.json())
except requests.exceptions.Timeout:
    print("요청 시간이 초과되었습니다.")
except requests.exceptions.HTTPError as error:
    print("HTTP 오류가 발생했습니다:", error)
except requests.exceptions.RequestException as error:
    print("요청 중 오류가 발생했습니다:", error)
```

```python
# 404 not found

error_url = "https://jsonplaceholder.typicode.com/posts/1111111111"

try:
    response = requests.get(error_url, timeout=5)

    # 400, 500번대도 무사히 클라이언트에 응답이 도착합니다.
    # 그 상황은 client의 입장에서는 data가 정상적으로 오지 않았기때문에 문제상황이라고 할 수 있습니다.
    response.raise_for_status()
    
    print("status code", response.status_code)
    
    print("요청 성공:", response.json())
except requests.exceptions.Timeout:
    print("요청 시간이 초과되었습니다.")
except requests.exceptions.HTTPError as error:
    print("HTTP 오류가 발생했습니다:", error)
except requests.exceptions.RequestException as error:
    print("요청 중 오류가 발생했습니다:", error)
```

```python
# time out

error_url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    response = requests.get(error_url, timeout=0.1)
    response = requests.get(error_url, timeout=5)

    # 400, 500번대도 무사히 클라이언트에 응답이 도착합니다.
    # 그 상황은 client의 입장에서는 data가 정상적으로 오지 않았기때문에 문제상황이라고 할 수 있습니다.
    response.raise_for_status()
    
    print("status code", response.status_code)
    print("요청 성공:", response.json())
except requests.exceptions.Timeout:
    print("요청 시간이 초과되었습니다.")
except requests.exceptions.HTTPError as error:
    print("HTTP 오류가 발생했습니다:", error)
except requests.exceptions.RequestException as error:
    print("요청 중 오류가 발생했습니다:", error)
```
