2. `시도별 실시간 측정정보 조회`의 `서울`의 데이터에 대해, 초 미세먼지 농도가 가장 낮은 `stationName`을 찾으시오.
    1. `시도별 실시간 측정정보 조회`의 `제주`의 데이터에 대해, 초 미세먼지 농도가 가장 낮은 `stationName`을 찾으시오.

```python
import time
from urllib.parse import unquote

def get_dust(sidoName='서울', attempt=3):
    for _ in range(attempt):
        try:
            url = "https://apis.data.go.kr/B552584/ArpltnInforInqireSvc/getCtprvnRltmMesureDnsty"

            params = {
                "serviceKey" : unquote(mi_api_key),
                "returnType" : 'json',
                "numOfRows" : 100,
                "pageNo" : 1,
                "sidoName" : sidoName,
                "ver" : 1.0
            }   
            response = requests.get(url, params=params)
            response.raise_for_status()
            json_data = response.json()
            response = json_data.get('response')
            body = response.get('body')
            items = body.get('items')
            return items
        except Exception as e :
            print(e)
            time.sleep(1)

data = get_dust()
# pprint(data)

def get_low_pm25_station(sidoName='서울'):
    return min(
        [x for x in get_dust(sidoName) if x.get("pm25Value") and x.get("pm25Value") != '-'], key=lambda x: int(x['pm25Value'])
    ) ['stationName']
    
pprint(get_low_pm25_station('서울'))
pprint(get_low_pm25_station('제주'))
```

3. `시도별 실시간 측정정보 조회`의 `서울`의 데이터를 `stationName`으로 접근하기 쉬운 자료구조로 재구성하시오. 


```python
def get_my_info(stationName) :
    my_info = {}
    data = get_dust()

    for station in data:
        station_name = station.get("stationName")
        # print(station_name)
        my_info[station_name] = station
        # print(my_info)
    return my_info[stationName]

pprint(get_my_info('관악구'))
```
