### 문제 1
현재 상영 중인 영화(Now Playing) 중 평점(vote_average)이 가장 높은 영화를 찾으시오.

```python
import requests
import os
from pprint import pprint
from dotenv import load_dotenv

load_dotenv()
url = "https://api.themoviedb.org/3/movie/now_playing"

api_key = os.getenv('api_key')

my_parameter = {
    "api_key" : api_key
}

try :
    response = requests.get(url, my_parameter)
    response.raise_for_status()
    data = response.json()
    # pprint(data)
    results = data.get('results')
    # pprint(results)

    best_movie = max(results, key=lambda x : x['vote_average'])
    print(best_movie['title'], best_movie['vote_average'])
except Exception as e :
    print(e)
```

### 문제 2
인기있는 영화(Popular) 중 평점(vote_average)이 가장 높은 영화를 찾으시오.

```python
def get_movies(subject="now_playing", page=1) :
    url = f"https://api.themoviedb.org/3/movie/{subject}"

    api_key = os.getenv('api_key')

    my_params = {
        'page' : page,
        "api_key" : api_key
    }

    response = requests.get(url, params=my_params)
    response.raise_for_status()
    data = response.json()
    # pprint(data)
    results = data.get('results')
    # pprint(results)
    return results

def get_best_movie(movies) :
    best_movie = max(movies, key=lambda x : x['vote_average'])
    return best_movie

try :
    moives = get_movies('popular')

    best_movie = get_best_movie(moives)

    print(best_movie['title'], best_movie['vote_average'])
except Exception as e :
    print(e)
```

### 문제 3
현재 상영 중인 영화(now_playing) 데이터에서 다음 정보만 담긴 리스트를 만드세요.

- title (영화 제목)  
- vote_average (평점)  
- release_date (개봉일)

```python
def get_simple_movie(movie) :
    title = movie["title"]
    vote_average = movie["vote_average"]
    release_date = movie["release_date"]

    simple_movie = {
        'title' : title,
        'vote_average' : vote_average,
        'release_date' : release_date
    }
    return simple_movie
try :
    moives = get_movies('now_playing')
    simple_movie = [get_simple_movie(movie) for movie in moives]

    pprint(simple_movie)
except Exception as e :
    print(e)
```

### 문제 4
현재 상영 중인 영화(now_playing) 데이터를 vote_average(평점)순으로 정렬하여 영화 제목이 담긴 리스트를 만드세요.
단, 리스트의 길이는 40 이상이어야 합니다.


```python
try :
    pages = [1, 2, 3, 4]
    many_movies = []

    for page in pages :
        movies = get_movies('now_playing', page = page)
        many_movies.extend(movies)

    sorted_movies = sorted(many_movies, key=lambda x : x['vote_average'], reverse=True)
    sorted_movies_title = [ (movie['title'], movie['vote_average'])for movie in sorted_movies]

    pprint(sorted_movies_title)
except Exception as e :
    print(e)
```

### 문제 8
현재 상영 중인 영화(now_playing) 데이터에서 다음 정보만 담긴 리스트를 만드세요.
단, 장르는 text로 이루어진 list입니다.

- title (영화 제목)  
- genres (장르)

hint) API 문서의 endpoint 항목을 참고하세요


```python
def get_movie_genres() :
    movie_list = []
    movies = get_movies()

    for movie in movies :
        movie_id = movie['id']
        movie_detail = get_movie_by_id(movie_id)

        genres = [g['name'] for g in movie_detail['genres']]

        movie_list.append({
            'title' : movie_detail['title'],
            'genres' : genres
        })
    return movie_list
try :
    result = get_movie_genres()
    pprint(result)
except Exception as e:
    print(e)
```
