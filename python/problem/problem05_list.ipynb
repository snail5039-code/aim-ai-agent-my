### 문제 9
1부터 50까지의 숫자 중에서 5의 배수만 골라 `my_list`에 저장하고, 그 리스트의 모든 요소의 합을 `for`문을 사용하여 구하세요.

```python
my_list = [i for i in range(1, 51) if i % 5 == 0]

result = 0

for i in my_list :
    result += i
print(result)
```

### 문제 11
사용자로부터 숫자를 계속 입력받다가 `-1`을 입력하면 입력을 멈추고, 그동안 입력받은 숫자들 중에서 **가장 큰 값**을 출력하는 프로그램을 `while`문을 사용하여 작성하세요. (단, `max()` 함수를 사용하지 마세요.)

```python
li = []
while True :
    input_int = int(input())
    li.append(input_int)

    if input_int == -1 :
        break
    
result = li[0]
for i in li :
    if result < i : 
        result = i
        
print(result)
```

### 문제 12
`while`문을 사용하여 1부터 100까지의 정수 중 홀수의 합과 짝수의 합을 각각 따로 구하여 출력하세요.

```python
num = 1
num1 = 0
num2 = 0

while num <= 100 :
    if num % 2 == 0 :
        num1 += num
    else :
        num2 += num
    num += 1
print(f"짝수 : {num1}, 홀수 : {num2}")
```
