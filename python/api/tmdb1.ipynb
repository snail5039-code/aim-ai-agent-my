### 문제 1

```python
url = "https://api.themoviedb.org/3/movie/now_playing"

token = os.getenv("token")

my_headers = {
    "Authorization" : f"Bearer {token}",
}


try:

    response = requests.get(url, headers=my_headers)

    response.raise_for_status()

    data = response.json()
    # pprint(data)
    results = data.get('results')
    # pprint(result)
    
    max_value = 0
    best_movie = ""
    for movie in results :
        # print(movie)

        vote_average = movie.get('vote_average')
        # print(vote_average)
        # print()
        if max_value < vote_average :
            max_value = vote_average
            best_movie = movie.get('title')
    print(best_movie, max_value)
    
except Exception as e:
    print('오류 발생!')
    print(e)
```

### 문제 2
인기있는 영화(Popular) 중 평점(vote_average)이 가장 높은 영화를 찾으시오.

```python
url = "https://api.themoviedb.org/3/movie/popular"

token = os.getenv("token")

my_headers = {
    "Authorization" : f"Bearer {token}",
}


try:
    response = requests.get(url, headers=my_headers)

    response.raise_for_status()

    data = response.json()
    # pprint(data)
    results = data.get('results')
    # pprint(result)
    
    best_movie = max(results, key=lambda x : x['vote_average'])
    print(best_movie['title'], best_movie['vote_average'])
    
except Exception as e:
    print('오류 발생!')
    print(e)
```

```python
# 1. 영화 데이터를 가져온다.
# 2. 영화 데이터를 바탕으로 평점이 가장 높은 영화를 찾는다.

# 함수의 매개변수 자리에 할당을 하면 "기본값"의 역할을 합니다.
# 값이 들어오지 않으면 기본값으로 할당되어 실행됩니다.

# page대신 params를 넣는 것이 더 확장성이 있긴 합니다.
def get_movies(what="now_playing", page=1):
    # movie/ 뒤에 now_playing or popular
    url = f"https://api.themoviedb.org/3/movie/{what}"

    token = os.getenv("token")

    my_params = {
        # language
        "page" : page
        # region
    }

    my_headers = {
        "Authorization" : f"Bearer {token}",
    }

    response = requests.get(url, headers=my_headers, params=my_params)

    # 함수 내부에서는 raise만 하고, except 처리는 밖에서 진행합니다.
    # 항상 그래야하는건 아닙니다.
    response.raise_for_status()

    data = response.json()
    
    results = data.get("results")
    return results

def get_best_movie(movies):
    
    best_movie = max(movies, key=lambda movie : movie['vote_average'])

    return best_movie


try:
    # 1. popular 영화 데이터 가져올꺼야.
    movies = get_movies('popular')
    # pprint(movies)

    # 2. 평점 높은 영화 가져올꺼야.
    best_movie = get_best_movie(movies)
    print(best_movie['title'], best_movie['vote_average'])
except Exception as e:
    print('오류 발생!')
    print(e)    

# print(best_movie)
```

### 문제 4
현재 상영 중인 영화(now_playing) 데이터를 vote_average(평점)순으로 정렬하여 영화 제목이 담긴 리스트를 만드세요.
단, 리스트의 길이는 20 이상이어야 합니다.


```python
url = "https://api.themoviedb.org/3/movie/now_playing"

token = os.getenv("token")

my_headers = {
    "Authorization" : f"Bearer {token}",
}
try:
    res1 = requests.get(url, headers=my_headers, params={"page": 1})  
    res1.raise_for_status()
    results1 = res1.json().get("results", []) 

    res2 = requests.get(url, headers=my_headers, params={"page": 2})
    res2.raise_for_status()  
    results2 = res2.json().get("results", [])  

    results = results1 + results2 
    
    lst = []

    for movie in results :
        
        title = movie.get('title')
        vote_average = movie.get('vote_average')
        lst.append([title, vote_average])
    lst.sort(key=lambda x : x[1], reverse=True)
    pprint(lst)

    
except Exception as e:
    print('오류 발생!')
    print(e)
```

```python
try:
    # 아래 로직을 함수로 만들어서
    # n페이지까지의 정보를 한번에 가져오는 함수를 만들수도 있을겁니다.
    pages = [1, 2, 3, 4]
    many_movies = []
    for page in pages:

        movies = get_movies(page=page)
        # many_movies += movies
        many_movies.extend(movies)
    # pprint([movie['title'] for movie in many_movies])
    # many_movies를 활용해서 vote_aveate 순으로 정렬 -> 영화 제목만 가져오기

    # sorted 
    # key 사용해서 기준 잡기
    # reverse=True -> 역순으로 정렬

    sorted_movies = sorted(many_movies, key=lambda movie : movie['vote_average'], reverse=True)

    sorted_movies_title = [(movie['title'], movie['vote_average']) for movie in sorted_movies]

    pprint(sorted_movies_title)

except Exception as e:
    print(e)
```

```python
import os  #
from pprint import pprint  
from dotenv import load_dotenv  
import requests 

load_dotenv()  
token = os.getenv("token")  

my_headers = {  
    "Authorization": f"Bearer {token}", 
}  

try:  
    genre_url = "https://api.themoviedb.org/3/genre/movie/list"  
    genre_res = requests.get(genre_url, headers=my_headers)  
    genre_res.raise_for_status()  

    genre_data = genre_res.json()  
    genres_list = genre_data.get("genres")
    # print(genres_list)

    genre_dict = {item['id'] : item['name'] for item in genres_list}
    # print(gnere_dict)

    now_playing_url = "https://api.themoviedb.org/3/movie/now_playing" 
    response = requests.get(now_playing_url, headers=my_headers)  
    response.raise_for_status()  

    data = response.json()  
    results = data.get("results")
    # print(results)
    final_list = []
    for movie in results :
        genre_name = [genre_dict[gid] for gid in movie.get("genre_ids") if gid in genre_dict]

        final_list.append(
                {  
                    "title": movie.get("title"),  
                    "genres": genre_name,  
                }
            )  

    pprint(final_list)  # 최종 결과 출력
except Exception as e:  
    print("오류 발생!")  
    print(e) 
```

```python

try:
    movies = get_movies()

    genres_data = fetch_tmdb_api("genre/movie/list")
    genres = genres_data.get('genres')
    my_movie_list = []

    # 얼마나 반복을 도는지 확인
    count = 0
    
    # 영화가 가지고 있는 장르 id를 활용해서
    # api에서 가져온 장르 id - name을 한번에 가져오고 싶다.
    genre_mapping = {
        "1" : "액션",
        "2" : "스릴러"
    }
    # 처럼 만들면 genre_mapping.get(1) => 액션이 O(1)으로 나온다.

    # genres => genre_mapping 만들어야 겠다.

    genre_mapping = {}

    for genre in genres:
        genre_id = genre.get('id')
        genre_name = genre.get('name')

        genre_mapping[genre_id] = genre_name

        count += 1
    pprint(genre_mapping)

    my_movie_list = []

    for movie in movies:
        movie_genre_ids = movie.get('genre_ids')

        movie_genre_list = []
        for genre_id in movie_genre_ids:
            genre_name = genre_mapping[genre_id]
            movie_genre_list.append(genre_name)

            count += 1
        my_movie_list.append((movie.get('title'), movie_genre_list))

    print(count)

    # api 실시간이여서 여러분들이 돌릴때는 다를수도 있습니다.
    # 위의 코드 break 없을 때 : 1102
    # 위의 코드 break 있을 때 : 429
    # 이번 코드              : 77
    pprint(my_movie_list)
    
        

except Exception as e:
    print('오류발생!')
    print(e)
```
