
### 문제 1
사용자로부터 정수 `n`을 입력받아, 1부터 `n`까지 모든 정수의 합을 출력하세요.

````python
num = int(input())

result = 0

for i in range(num + 1) :
    result += i
print(result)
````

````python
n = 20

total = 0
num = 1

while num <= 20 :
    total += num
    num += 1
print(total)
````

### 문제 2

지정된 비밀번호(예: `"python123"`)가 입력될 때까지 계속해서 입력을 받는 프로그램을 작성하세요. 틀릴 경우 "다시 시도하세요"를 출력합니다.

````python
word = input()
while word != "python123" :
    print("다시 시도하세요")
    word = input()
````

````python
valid_pw = 'python123'

while True :
    input_pw = input()

    if input_pw == valid_pw :
        break
    else :
        print('다시 시도하세요')
````

### 문제 3

1부터 30까지의 숫자 중에서 3의 배수를 제외한 나머지 숫자들만 출력하세요.

````python
for i in range(1, 31) :
    if i % 3 != 0 :
        print(i)
````

````python
i = 1
while i <= 30 :
    if i % 3 != 0 :
        print(i)
    i += 1
````

### 문제 4
`names = ['jun', 'ken', 'alex', 'sophia', 'neo']` 일 때,

출석번호를 포함하여 '1번 jun', ..., '5번 neo'의 형태로 출석을 부르시오.

````python
names = ['jun', 'ken', 'alex', 'sophia', 'neo']

for index, name in enumerate(names) :
        print(f"{index + 1}번 {name}")
        index += 1

for index in range(len(names)) :
        name = names[index]
        print(index + 1, name)
````

### 문제 5
```
names = ['jun', 'alex', 'ken']
ages = [18, 28, 37]
``` 
일 때,
```
jun 18
alex 28
ken 37
```
을 출력하시오.

````python
names = ['jun', 'alex', 'ken']
ages = [18, 28, 37]

for index in range(len(names)) :
    name = names[index]
    age = ages[index]
    print(name, age)

for name, age in zip(names, ages) :
    print(name, age)
````

### 문제 6

사용자로부터 숫자를 계속 입력받다가, `0`을 입력하면 지금까지 입력된 모든 숫자의 합계를 출력하고 종료하는 프로그램을 작성하세요.

````python
result = 0
while True :
    input_user = int(input())
    if input_user == 0 :
        break
    result += input_user
print(result)

# for문은 지금 문제에는 적합하지 않다
````

### 문제 7

1부터 20까지의 숫자 중 홀수는 건너뛰고, 짝수들만 더하다가 합계가 40을 넘으면 멈추고 합계를 출력하는 프로그램을 작성하세요.

````python
result = 0
for num in range(1, 21) :
    if num % 2 == 0 :
        result += num
        if result > 40 :
            break
print(result)
````

````python
result = 0
num = 1

while num <= 20 :
    if num % 2 == 0 :
        result += num
        if result > 40 :
            break
    num += 1
print(result)    
````
