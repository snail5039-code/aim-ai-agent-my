### 문제
2와 근접해있는 1의 개수 구하기 => 2와 근접 : 상하좌우 중 하나가 맞닿아 있음

```python
# 5 5
# 1 0 1 0 2
# 0 0 1 2 1
# 0 1 0 1 2
# 1 1 0 2 1
# 2 0 1 0 2
n, m = 5, 5
lst = [
    [1, 0, 1, 0, 2], 
    [0, 0, 1, 2, 1], 
    [0, 1, 0, 1, 2], 
    [1, 1, 0, 2, 1], 
    [2, 0, 1, 0, 2]
    ]

di = [0, 0, 1, -1]
dj = [1, -1, 0, 0]

count = 0

for i in range(n) :
    for j in range(m) :
        if lst[i][j] == 2 :
            for k in range(4) :
                ni = i + di[k]
                nj = j + dj[k]

                if 0 <= ni < n and 0 <= nj < m :
                    if lst[ni][nj] == 1 :
                        count += 1
                        lst[ni][nj] = 0
print(count)
```
