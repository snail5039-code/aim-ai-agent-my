```python
from abc import ABC, abstractmethod

class SmartHub:
    def __init__(self, name, protocol):
        self.name = name
        self.protocol = protocol
        self.devices = []
    def register(self, device):
        if type(device.protocol) == type(self.protocol):
            self.devices.append(device)
            print(f"[{self.name}] '{device.name}' 등록 성공")
        else : 
            print("호완 안됌")
    def start_all(self):
        print(f"{self.name}전체 가동")
        for device in self.devices:
                device.func()
    def stop_all(self):
        print(f"{self.name}전체 중단")
        for device in self.devices:
                device.stop()


class Device(ABC):
    def __init__(self, name, company, protocol):
        self.name = name
        self.company = company
        self.protocol = protocol
    @abstractmethod
    def func(self):
        pass
    @abstractmethod
    def check_status(self):
        pass
    @abstractmethod
    def stop(self):
        pass

class SmartLight(Device):
    def func(self):
        self.protocol.protocol_start()
        print(f"{self.name} 작동")
    def check_status(self):
        print(f"{self.name} 상태 확인")
    def stop(self):
        print(f"{self.name} 작동 중단")

class AirConditioner(Device):
    def func(self):
        self.protocol.protocol_start()
        print(f"{self.name} 작동")
    def check_status(self):
        print(f"{self.name} 상태 확인")
    def stop(self):
        print(f"{self.name} 작동 중단")


class Protocol(ABC):
    @abstractmethod
    def protocol_start(self):
        pass

class WiFiProtocol(Protocol):
    def protocol_start(self):
        print("WiFi 연결을 시도합니다")
class ZigbeeProtocol(Protocol):
    def protocol_start(self):
        print("Zigbee망을 구성합니다")
class BluetoothProtocol(Protocol):
    def protocol_start(self):
        print("블루투스 연결을 시도합니다")

xiaomi_hub = SmartHub("샤오미 허브", ZigbeeProtocol())
philips_light = SmartLight("필립스 전등", "필립스", WiFiProtocol())
ikea_light = SmartLight("이케아 전등", "이케아", ZigbeeProtocol())

print("=== 등록 테스트 ===")
xiaomi_hub.register(philips_light)  # 실패 (WiFi != Zigbee)
xiaomi_hub.register(ikea_light)     # 성공 (Zigbee == Zigbee)

print("\n=== 가동 테스트 ===")
xiaomi_hub.start_all()  # 전체 가동 (Zigbee망 구성 -> 이케아 전등 작동)

print("\n=== 상태 확인 테스트 ===")
# 방법 A: 기기 개별 직접 확인
ikea_light.check_status()  

# 방법 B: 허브에 check_all()을 만들었을 경우
# xiaomi_hub.check_all()

print("\n=== 중단 테스트 ===")
xiaomi_hub.stop_all()   # 전체 중단
```
