### 문제 1
입력받은 점수에 따라 합격/불합격을 판별하고 메시지를 반환하는 함수 `get_grade_result`를 작성하세요.

**조건:**
1. 점수가 60점 이상이면 "합격"을 반환합니다.
2. 점수가 60점 미만이면 "불합격"을 반환합니다.
3. 만약 점수가 0점보다 작거나 100점보다 크면 "잘못된 점수"를 반환하세요.

```python
def get_grade_result(score):
    if score < 0 or score > 100 :
        return "잘못된 점수"
    elif score >= 60 :
        return "합격"
    else :
        return "불합격"

get_grade_result(80)
```

```python


# 테스트
print(get_grade_result(85))  # 결과: "합격"
print(get_grade_result(40))  # 결과: "불합격"
print(get_grade_result(120)) # 결과: "잘못된 점수"
```

### 문제 3
두 개의 숫자와 연산자(+, -)를 입력받아 계산 결과를 반환하는 간단한 계산기 함수 `simple_calc`를 작성하세요.

**조건:**
1. `op`가 "+"이면 두 수의 합을 반환합니다.
2. `op`가 "-"이면 두 수의 차를 반환합니다.
3. 그 외의 연산자가 들어오면 "지원하지 않는 연산"이라는 메시지를 반환하세요.

```python
num1 = 10
num2 = 20
op = '+'

def simple_calc(num1, num2, op):
    if op == '+' :
        return num1 + num2
    elif op == "-" :
        if num1 < num2 :
            return num2 - num1
        return num1 - num2
    else :
        return "지원하지 않는 연산"
```

```python

# 테스트
print(simple_calc(10, 5, "+")) # 결과: 15
print(simple_calc(10, 5, "-")) # 결과: 5
print(simple_calc(10, 5, "*")) # 결과: "지원하지 않는 연산"
```

### 문제 4
이름 리스트를 입력받아 이름이 3자 이상인 사람만 골라 새로운 리스트로 반환하는 함수 `filter_long_names`를 작성하세요.

**조건:**
1. `for`문을 사용하여 리스트의 각 이름을 확인합니다.
2. `len()` 함수를 사용하여 이름의 길이를 확인합니다.

```python
name_list = ["김철수", "이영희", "박씨", "최도날드"]

def filter_long_names(name_list):
    result = []
    for name in name_list :
        if len(name) >= 3 :
            result.append(name)
    return result 

```

```python

# 테스트
name_list = ["김철수", "이영희", "박씨", "최도날드"]
print(filter_long_names(name_list)) # 결과: ["김철수", "이영희", "최도날드"]
# print([name for name in name_list if len(name) >= 3])
```

```python
username = 'abcd1234'
age = 20
email = "admin@example.com"
def validate_user_registration(username, age, email):
    result = False
    if len(username) >= 4 :
        result = True
        if age >= 19 :
            result = True
            if '@' in email :
                result = True
    return result
# 이렇게 실제 서비스에 입력받으면 위쪽 부터 통과해야한다
```

### 문제 6
장바구니에 담긴 상품들의 최종 결제 금액을 계산하는 함수 `calculate_checkout_price`를 작성하세요.

**조건:**
1. `membership_grade`가 'VIP'이면 총액의 15%, 'GOLD'이면 10%를 할인합니다.
2. 할인 적용 후 최종 금액이 50,000원 미만이면 배송비 3,000원을 추가합니다.
3. 50,000원 이상이면 배송비는 무료(0원)입니다.

```python
my_cart = [{
    "price": 25000, 
    "quantity": 1
    }, {"price": 15000, "quantity": 2}]


def calculate_checkout_price(my_cart, membership_grade):
    total_price = 0
    for cart in my_cart :
        total_price += cart['price'] * cart['quantity']
    if membership_grade == 'VIP' :
        total_price = total_price * 0.85
    elif membership_grade == "GOLD" :
        total_price = total_price * 0.9
    if total_price < 50000 :
        total_price += 3000
    return round(total_price)
```

```python
# 유지보수에 용이하게 만들기 기능별로 분리 시킨다음 최종 함수에서 실행 시킬 수 있게!

my_cart = [{
        "price": 25000, 
        "quantity": 1
    }, {"price": 15000, "quantity": 2}]


def calculate_discount_rate(membership_grade):
    # membership_grade를 가지고 discount_rate를 알려주는 함수.
    # 할인률 계산
    discount_rate_mapping = {
        'VIP' : 0.15,
        'GOLD' : 0.1,
    }
    return discount_rate_mapping.get(membership_grade, 0)

def calculate_shipping_fee(total):
    shipping_fee = 3000
    base_price = 50000

    return 0 if total >= base_price else shipping_fee

def calculate_total_price(my_cart) :
    total = 0
    for product in my_cart:
        price = product['price']
        quantity = product['quantity']
        total += price * quantity
    return total

def calculate_checkout_price(my_cart, membership_grade):

    # 물건 총 합계 계산
    total = calculate_total_price(my_cart)

    # 할인 적용
    discount_rate = calculate_discount_rate(membership_grade)
    total = total * (1 - discount_rate)

    # 배송비 적용
    total = total + calculate_shipping_fee(total)

    return int(total)
```

```python
def calculate_checkout_price(my_cart, membership_grade):

    # 물건 총 합계 계산
    total = calculate_total_price(my_cart)

    # 할인 전 배송비 계산
    shipping_fee = calculate_shipping_fee(total)

    # 할인 적용
    discount_rate = calculate_discount_rate(membership_grade)
    total = total * (1 - discount_rate)

    # 할인 전 금액 기준의 배송비 적용 - 정답보다 3000원 적다.
    total = total + shipping_fee

    return int(total)
```

**조건:**
1. 숫자형 문자열("105"), 숫자(101), 리스트 안의 숫자([102]) 등을 모두 정수로 변환하세요.
2. 중복된 ID는 제거하고, 결과는 오름차순으로 정렬된 리스트여야 합니다.

```python
dirty_data = ["105", 101, [102, 103], "101", 104, [105]]

# dirty_data = ["105", 101, "101", 104, ]

def normalize_user_data(dirty_data):
    data_li = []
    for data in dirty_data :
        if isinstance(data, list) :
            for item in data :
                item_num = int(item)
                if item_num not in data_li :
                    data_li.append(int(item))
        else :
            item_num = int(data)
            if item_num not in data_li :
                data_li.append(int(data))
    data_li.sort()
    return data_li

```

```python
def normalize_user_data(dirty_data) :
    clean_data = []
    for data in dirty_data :
        if type(data) == list :
            for datum in data :
                clean_data.append(datum)
        else :
            clean_data.append(int(data))

    clean_data = set(clean_data) # 중복제거
    clean_data = sorted(clean_data)
    clean_data = list(clean_data)
    return clean_data
```

```python


dirty_data = ["105", 101, [102, 103], "101", 104, [105]]
print(normalize_user_data(dirty_data)) # 결과: [101, 102, 103, 104, 105]
```

### 문제 8
특정 조건을 만족하는 유저들을 추출하여 알림 대상자 리스트를 반환하는 함수 `get_notification_targets`를 작성하세요.

**조건:**
1. `is_active`가 `True`인 유저 중에서 선별합니다.
2. `last_login`이 7일 이상이거나, `point`가 10,000점 이상인 유저가 대상입니다.
3. 대상자의 'name'만 리스트에 담아 반환하세요.

```python
users_data = {
    "u1": {"name": "Alice", "is_active": True, "last_login": 10, "point": 5000},
    "u2": {"name": "Bob", "is_active": False, "last_login": 1, "point": 15000},
    "u3": {"name": "Charlie", "is_active": True, "last_login": 2, "point": 12000}
}



def get_notification_targets(users_data):
    li = []
    for user in users_data.values() :
        if user['is_active'] == True and (user['last_login'] >= 7 or user['point'] >= 10000) :
            li.append(user['name'])
    return li

def get_notification_targets(users_data):
    li = []
    for user in users_data.values() :
        if not user['is_active'] :
            continue
        if not (user['last_login'] >= 7 or user['point'] >= 10000) :
            continue
        li.append(user['name'])
    return li
```

```python


users_data = {
    "u1": {"name": "Alice", "is_active": True, "last_login": 10, "point": 5000},
    "u2": {"name": "Bob", "is_active": False, "last_login": 1, "point": 1500},
    "u3": {"name": "Charlie", "is_active": True, "last_login": 2, "point": 12000}
}
print(get_notification_targets(users_data)) # 결과: ['Alice', 'Charlie']
```
